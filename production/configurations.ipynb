```python
def extract_track_id(uri: str) -> str | None:
    """Pulls the ID out of a URI like 'spotify:track:3n3Ppam7vgaVa1iaRUIOKE'."""
    if uri and uri.startswith("spotify:track:"):
        return uri.split(":")[-1]
    return None  # local files or null URIs
```
