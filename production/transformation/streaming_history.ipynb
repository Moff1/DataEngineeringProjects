```python
import glob
import json
import os
import pandas as pd
```

```python
DIRECTORY   = 'C:\\Users\\JosephMoffatt\\OneDrive - Thought Logic\\Desktop\\GitHubRepos\\dataengineeringprojects\\DataEngineeringProjects\\auxdata\\'   # <-- change this
OUTPUT_FILE = "combined_streaming_history.json"
```

```python
def find_files(directory: str) -> list[str]:
    """Return all files matching StreamingHistory_music* in the given directory."""
    pattern = os.path.join(directory, "StreamingHistory_music*")
    return sorted(glob.glob(pattern))
```

```python
def load_and_combine(files: list[str]) -> list[dict]:
    """Load each JSON file and combine all records into a single list."""
    combined = []
    for filepath in files:
        filename = os.path.basename(filepath)
        try:
            with open(filepath, "r", encoding="utf-8") as f:
                data = json.load(f)
            if isinstance(data, list):
                combined.extend(data)
                print(f"  [OK] {filename} — {len(data):,} records")
            else:
                print(f"  [SKIP] {filename} — unexpected format (not a list)")
        except json.JSONDecodeError as e:
            print(f"  [ERROR] {filename} — JSON parse error: {e}")
        except OSError as e:
            print(f"  [ERROR] {filename} — could not read file: {e}")
    return combined
```

```python
def save_output(records: list[dict], output_path: str) -> None:
    """Write the combined records to a JSON file."""
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(records, f, ensure_ascii=False, indent=2)
    print(f"\nSaved {len(records):,} total records to: {output_path}")
```

```python
def main():
    print(f"Scanning: {DIRECTORY}\n")
 
    files = find_files(DIRECTORY)
 
    if not files:
        print("No files found matching 'StreamingHistory_music*'.")
        return
 
    print(f"Found {len(files)} matching file(s):\n")
    combined = load_and_combine(files)
 
    output_path = os.path.join(DIRECTORY, OUTPUT_FILE)
    save_output(combined, output_path)
 
    # Quick summary
    print("\n── Summary ──────────────────────────────")
    print(f"  Files processed : {len(files)}")
    print(f"  Total records   : {len(combined):,}")
    print(f"  Output file     : {output_path}")
 
 
if __name__ == "__main__":
    main()
```
